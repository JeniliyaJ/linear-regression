```python
##select the feature
```

```python
x=ad_sales['tv']
y=ad_sales['sales']
```

```python
x.shape
```

```python
#convert 1d to 2d
```

```python
x=np.reshape(x,shape=(-1,1))
```

```python
x.shape
```

```python
#linear regression using linear_model
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,
    train_size=0.7, random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
#train the model
```

```python
lm= LinearRegression()
```

```python
lm.fit(x_train,y_train)
```

```python
lm.coef_
```

```python
lm.intercept_
```

```python
#predict the model
x_test
```

```python
y_pred=lm.predict(x_test)
```

```python
y_pred
```

```python
new_data=lm.predict([[300]])
new_data
```

```python
#evaluate the error in my model
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
r2_score(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

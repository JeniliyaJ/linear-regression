```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
hours = np.array([[1],[2],[3],[4],[5]]) #input is always should be a 1D
marks = np.array([35,40,45,50,55])
```

```python
#creating the model
model=LinearRegression()
```

```python
# train the model
model.fit(hours,marks)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
predicted=model.predict([[6],[7],[8]])
```

```python
print(predicted)
```

```python
model_predict=model.predict(hours)
```

```python
print(model_predict)
```

```python
##visualizing regression
```

```python
plt.scatter(hours,marks)
plt.plot(hours,model_predict,color='r') ## preduicted line
plt.scatter([6,7,8],predicted,color='green')
```

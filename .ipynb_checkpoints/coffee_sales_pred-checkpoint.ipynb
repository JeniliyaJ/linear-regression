## preprocessing method


```python
encoding= LabelEncoder()
```

```python
coffee_sales['Day_Name']=encoding.fit_transform(coffee_sales['Day_Name'])
```

```python
coffee_sales['Season']=encoding.fit_transform(coffee_sales['Season'])
```

```python
coffee_sales.info()
```

## feature selection for modeling

```python
corr_col=coffee_sales.corr()['Daily_Revenue'].sort_values(ascending=False).head(15)
corr_col
```

```python
include_column=['Ingredient_Cost','Coffee_Sales','Num_Customers','Daily_Profit','Pastry_Sales','Total_Costs','Sandwich_Sales','Nearby_Events','Customer_Satisfaction','Is_Weekend','Day_of_Week','Temperature_C','Staff_Count','Staff_Cost']
```

```python
coffee_sales.corr()
```

```python
feature_col=[
    'Day_of_Week',
    'Month',
    'Season',
    'Temperature_C',
    'Is_Raining',
    'Rainfall_mm',
    'Is_Holiday',
    'Promotion_Active',
    'Nearby_Events',
    'Staff_Count',
    'Machine_Issues',
    'Num_Customers',
    'Coffee_Sales',
    'Pastry_Sales',
    'Sandwich_Sales'
]
```

```python
feature_col
```

```python
x=coffee_sales[feature_col]
y=coffee_sales['Daily_Revenue']
```

```python
x.shape
```

## split the data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=45)
```

```python
## scaling the data because our model is automatically considered large number data feature is used to prediction 
```

```python
scaling=StandardScaler()
```

```python
x_train_scale=scaling.fit_transform(x_train)
x_test_scale=scaling.transform(x_test)
```

```python
x_train_scale
```

## model creation

```python
coff_sales_pred=LinearRegression()
```

```python
coff_sales_pred.fit(x_train,y_train)
```

```python
y_pred=coff_sales_pred.predict(x_test)
y_train_pred=coff_sales_pred.predict(x_train)
```

## evaluate the model

```python
test_square=r2_score(y_test,y_pred)
test_square
```

```python
train_square=r2_score(y_train,y_train_pred)
train_square
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Daily Revenue")
plt.ylabel("Predicted Daily Revenue")
plt.title("Actual vs Predicted Daily Revenue")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(y_test.values, label="Actual")
plt.plot(y_pred, label="Predicted")

plt.xlabel("Test Data")
plt.ylabel("Daily Revenue")
plt.title("Actual vs Predicted Revenue")

plt.legend()
plt.show()
```
